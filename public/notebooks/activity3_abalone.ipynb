교과서 104~105쪽

전복 껍데기의 고리 수(rings)를 세면 나이(고리 수 + 1.5)를 추정할 수 있다. 전복의 크기와 무게로 고리 수를 예측하는 선형 회귀 모델을 만들어 보자.

**데이터**: 교과서 104쪽 링크(캐글 abalone-uci)의 `abalone_original.csv` — sex, length, diameter, height, whole-weight, shucked-weight, viscera-weight, shell-weight, rings

```python
#@title 데이터 준비하기
how = "사이트에서 바로 받기" #@param ["사이트에서 바로 받기", "내 컴퓨터에서 올리기"]
#@markdown 디지털 교과서에 올려 둔 교과서 자료(abalone_original.csv)를 코랩으로 바로 가져옵니다. 직접 모은 파일을 쓰려면 '내 컴퓨터에서 올리기'를 고르세요.
if how.startswith('사이트'):
    import urllib.request
    for name in ['abalone_original.csv']:
        urllib.request.urlretrieve('https://aibook-dl.vercel.app/data/' + name, name)
        print('받았습니다:', name)
else:
    from google.colab import files
    uploaded = files.upload()
```

```python
#@title ① 데이터 불러오고 살펴보기
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('abalone_original.csv')
df.info()
show_pairplot = False #@param {type:"boolean"}
if show_pairplot:
    sns.pairplot(df); plt.show()        # 시간이 조금 걸립니다
df.head()
```

```python
#@title ② 종속 변수를 고르고 선형 회귀 모델 만들기
target = "rings" #@param ["rings", "shucked-weight", "whole-weight"]
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
num = df.select_dtypes('number').columns.tolist()
drop = {'rings': [], 'shucked-weight': ['whole-weight', 'shell-weight'], 'whole-weight': ['shucked-weight', 'viscera-weight', 'shell-weight']}[target]
features = [c for c in num if c != target and c not in drop]
X, y = df[features], df[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model = LinearRegression().fit(X_train, y_train)
print('독립 변수:', features)
print('회귀계수:', model.coef_.round(3), ' 절편:', round(model.intercept_, 3))
print(f'테스트 결정계수(R²): {model.score(X_test, y_test):.3f}')
```
