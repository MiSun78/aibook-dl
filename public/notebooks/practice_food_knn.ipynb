k-최근접 이웃(kNN) 알고리즘으로 당도(sweet)와 아삭함(crunchy)을 보고 과일(fruit)인지 채소(vegetable)인지 분류해 보자.

**데이터**: 교과서 109쪽 링크의 `food.csv`. 파일이 없으면 아래 칸이 연습용 데이터를 만들어 씁니다.

```python
#@title 데이터 준비하기
how = "사이트에서 바로 받기" #@param ["사이트에서 바로 받기", "내 컴퓨터에서 올리기"]
#@markdown 디지털 교과서에 올려 둔 교과서 자료(food.csv)를 코랩으로 바로 가져옵니다. 직접 모은 파일을 쓰려면 '내 컴퓨터에서 올리기'를 고르세요.
if how.startswith('사이트'):
    import urllib.request
    for name in ['food.csv']:
        urllib.request.urlretrieve('https://aibook-dl.vercel.app/data/' + name, name)
        print('받았습니다:', name)
else:
    from google.colab import files
    uploaded = files.upload()
```

```python
#@title 데이터 불러오고 산점도 그리기
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
if os.path.exists('food.csv'):
    df = pd.read_csv('food.csv')
else:
    print('food.csv가 없어 연습용 데이터를 씁니다.')
    df = pd.DataFrame({
        'sweet':   [10, 3, 7, 9, 8, 6, 9, 2, 1, 3, 2, 4, 1, 8, 7],
        'crunchy': [1, 8, 2, 3, 6, 3, 4, 9, 7, 5, 6, 8, 4, 2, 5],
        'class':   ['fruit', 'vegetable', 'fruit', 'fruit', 'fruit', 'fruit', 'fruit', 'vegetable', 'vegetable',
                    'vegetable', 'vegetable', 'vegetable', 'vegetable', 'fruit', 'fruit']})
sns.scatterplot(x='sweet', y='crunchy', data=df, hue='class', s=80)
plt.show()
df.head()
```

```python
#@title k-최근접 이웃으로 분류하기
k = 3 #@param {type:"slider", min:1, max:9, step:2}
sweet = 2 #@param {type:"slider", min:0, max:10, step:1}
crunchy = 5 #@param {type:"slider", min:0, max:10, step:1}
from sklearn.neighbors import KNeighborsClassifier
X = df[['sweet', 'crunchy']].values         # 특징
y = df['class']                             # 타깃
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X, y)
print(f'당도 {sweet}, 아삭함 {crunchy} → {knn.predict([[sweet, crunchy]])[0]}')
dist, idx = knn.kneighbors([[sweet, crunchy]])
print('가장 가까운 이웃:'); print(df.iloc[idx[0]])
```
