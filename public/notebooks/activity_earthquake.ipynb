```python
#@title ② 데이터 불러오고 결측치 확인하기
import pandas as pd
df = pd.read_csv('earthquake.csv')
print(df.shape)
print(df.isnull().sum())
df.head()
```

```python
#@title ③ 결측치 행 삭제하고 '연도' 속성 만들기
df.dropna(axis=0, inplace=True)                         # 결측치가 있는 행 삭제
df['연도'] = pd.to_datetime(df['시간']).dt.year          # 시간에서 연도만 뽑기
print('남은 데이터:', df.shape)
df.head()
```

```python
#@title ④ 연도에 따른 지진 규모 산점도
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(x='연도', y='규모', data=df)
plt.show()
big = df.loc[df['규모'].idxmax()]
print(f"가장 큰 지진: {big['시간']}, 규모 {big['규모']}, {big['위치']}")
```

```python
#@title ⑤ 지역별 지진 발생 횟수 원그래프
df['지역'] = df['위치'].str[:2]          # '위치'의 앞 두 글자(경북, 전남 등)를 '지역'으로
loc = df['지역'].value_counts()           # 지역별 횟수
top_n = 8 #@param {type:"slider", min:3, max:15, step:1}
show = pd.concat([loc[:top_n], pd.Series({'기타': loc[top_n:].sum()})]) if len(loc) > top_n else loc
plt.pie(show, labels=show.index, autopct='%1.1f%%')
plt.show()
print(loc.head(top_n))
```
