설탕과 카페인 함량으로 카페 음료를 군집화하고, 사용자 취향에 맞는 군집의 음료를 추천해 보자.

**데이터**: 교과서 124쪽 링크의 `caffe_menu.csv` (name, sugars(g), caffeine(mg), kcal 등)

```python
#@title 데이터 파일 올리기
#@markdown 왼쪽 [파일] 아이콘에서 직접 올렸다면 이 칸은 건너뛰어도 됩니다.
from google.colab import files
uploaded = files.upload()
```

```python
#@title ① 특징 고르고 정규화한 뒤 군집 모델 만들기
x_attr = "sugars(g)" #@param ["sugars(g)", "caffeine(mg)", "kcal"]
y_attr = "caffeine(mg)" #@param ["sugars(g)", "caffeine(mg)", "kcal"]
k = 4 #@param {type:"slider", min:2, max:8, step:1}
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
df = pd.read_csv('caffe_menu.csv')
data = df[[x_attr, y_attr]]
data_scaled = MinMaxScaler().fit_transform(data)      # 범위가 달라서 0~1로 정규화
model = KMeans(n_clusters=k, n_init=10, random_state=42).fit(data_scaled)
df['cluster'] = model.labels_
plt.scatter(df[x_attr], df[y_attr], c=df['cluster'], cmap='tab10', s=50)
plt.xlabel(x_attr); plt.ylabel(y_attr); plt.title('Clustering of Cafe Drinks'); plt.show()
```

```python
#@title ② 군집별 음료 보기
cluster_no = 0 #@param {type:"slider", min:0, max:7, step:1}
print(f'{cluster_no}번 군집의 평균:', df[df['cluster'] == cluster_no][[x_attr, y_attr]].mean().round(1).to_dict())
df[df['cluster'] == cluster_no]['name'].head(20)
```
